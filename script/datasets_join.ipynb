```python
import pandas as pd

# file paths
albums_path = "../data/best_selling_albums.csv"
awards_path = "../data/artist_awards.csv"
output_path = "../data/best_selling_albums_with_awards.csv"

# load datasets
albums_df = pd.read_csv(albums_path)
awards_df = pd.read_csv(awards_path)

# merge on Artist
merged_df = albums_df.merge(awards_df, on="Artist", how="outer")

# check NA values after merge
award_columns = [
    "Grammy_Awards",
    "Billboard_Top_100_Hits",
    "Years_Active",
    "Record_Label",
]
na_summary = merged_df.isna().sum()

print("Albums rows:", len(albums_df))
print("Awards rows:", len(awards_df))
print("Merged rows:", len(merged_df))
print("\nNA counts in joined award columns:")
print(na_summary)

# save merged dataset
merged_df.to_csv(output_path, index=False)

merged_df.tail()
```

## Merge Choice and NA explanation

I used an outer join because I wanted to see data from both datasets and not just matching Artists from each dataset

- In the `artist_awards.csv`, Claude Sonnet 4.5 provided me with 3 extra artists (Steve Wonder, Marvin Gaye, Drake) that are not in `best_selling_albums.csv` for me to experiment with all the joins.
- Matching Artists bring in award fields from `artist_awards.csv`
- 3 non matching Artists created N/A values in the columns that belong to `best_selling_albums.csv`
